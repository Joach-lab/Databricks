```sql
CREATE OR REPLACE TABLE silver_lemonade_spend AS
SELECT
--create Table
    TO_DATE(TRIM(Date), 'M/d/yyyy') AS transaction_date,
    UPPER(TRIM(Transaction_Type)) AS transaction_type,
    TRIM(Item_Description) AS item_description,
    CAST(Quantity AS INT) AS quantity,
    CAST(Unit_Price_USD AS DECIMAL(10, 2)) AS unit_price_usd,
    CAST(Total_Amount_USD AS DECIMAL(10, 2)) AS total_amount_usd
FROM default.lemonade_raw_spend_csv;
```
