```sql
CREATE OR REPLACE TABLE silver_lemonade_spend_v2 AS
SELECT
    UPPER(TRIM(Transaction_ID)) AS transaction_id,
    TO_DATE(TRIM(Date_Posted), 'yyyy-MM-dd') AS date_posted,
    CASE
        WHEN LOWER(TRIM(Item_Category)) IN ('raw_materials', 'raw materials') THEN 'Raw_Materials'
        WHEN LOWER(TRIM(Item_Category)) = 'marketing' THEN 'Marketing'
        ELSE 'Misc'
    END AS category,
    CASE
        WHEN LOWER(TRIM(Item_Category)) IN ('raw_materials', 'raw materials') THEN 'COGS'
        ELSE 'OPEX'
    END AS p_l_bucket,
    TRIM(Vendor_Name) AS vendor_name,
    CAST(
        NULLIF(REGEXP_REPLACE(COALESCE(Amount_USD, ''), '[^0-9.]', ''), '')
        AS DECIMAL(10, 2)
    ) AS clean_amount_usd,
    TRIM(Payment_Terms) AS payment_terms
FROM default.lemonade_raw_sales_csv
WHERE Transaction_ID IS NOT NULL
  AND TRIM(Transaction_ID) != '';
```
